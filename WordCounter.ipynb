```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df = pd.read_csv('products.csv')
```

```python
df.head()
```

```python
vec = CountVectorizer(ngram_range=(1, 1), stop_words="english", min_df=1, binary=True)
X = vec.fit_transform(df["description"].dropna())

freq = pd.DataFrame({
    "phrase": vec.get_feature_names_out(),
    "count": X.sum(axis=0).A1,
}).sort_values("count", ascending=False)
```

```python
print(freq.head(30))
```

```python
#write results to file
filename = "words_count.csv"
freq.to_csv(filename, index=False)
```
